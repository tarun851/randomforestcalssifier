```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
import seaborn as sns
df=sns.load_dataset("titanic")
df.head()
```

```python
df.describe()
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
numcol=df.select_dtypes(include="number").columns
```

```python
df[numcol]=df[numcol].fillna(df[numcol].median())
```

```python
df.isnull().sum()
```

```python
objcol = df.select_dtypes(include="object").columns
```

```python
objcol
```

```python
for col in objcol:
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
catcol = df.select_dtypes(include="category").columns
```

```python
for col in catcol:
    df[col]=df[col].fillna(df[col].mode()[0])
df
```

```python


df = pd.get_dummies(df, drop_first=True)
df
```

```python
x=df.drop("alive_yes",axis=1)
```

```python
y=df["alive_yes"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=RandomForestClassifier()
```

```python
params={
    'n_estimators':[100,200,300],
    'max_depth':[5,7,9,11],
    'min_samples_leaf':[2,4,6,8],
    'min_samples_split':[2,4,5]
}
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(param_grid=params,estimator=model,cv=5,n_jobs=-1,scoring='accuracy')
grid.fit(x_train,y_train)
best_params=grid.best_params_
best_model=grid.best_estimator_
print(best_params)
print(best_model)
```
